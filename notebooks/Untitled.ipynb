```python
import matplotlib.pyplot as plt
import numpy as np

# Configuration du style
plt.style.use(
    "seaborn-v0_8-whitegrid"
    if "seaborn-v0_8-whitegrid" in plt.style.available
    else "default"
)

fig, ax1 = plt.subplots(figsize=(12, 6))

# -------------------------------------------------------------------------
# Graphique unique : Effectifs bruts par ethnie (BCSC v2)
# -------------------------------------------------------------------------
ethnies = [
    "Européenne\n(White)",
    "Asiatique\n(Asian)",
    "Hispanique",
    "Afro-Américaine\n(Black)",
    "Autres /\nInconnu",
]

# Effectifs bruts réels (Tice et al., 2015)
effectifs_totaux = [901975, 95435, 79141, 60519, 26769]  # Total = 1 163 839
cas_cancers = [14857, 1178, 1035, 1002, 296]  # Total = 18 368

x = np.arange(len(ethnies))
width = 0.35

# Axe principal (Gauche) : Nombre total de femmes incluses
rects1 = ax1.bar(
    x - width / 2,
    effectifs_totaux,
    width,
    label="Femmes incluses (Total N=1 163 839)",
    color="#000",
    edgecolor="black",
    alpha=0.85,
)
ax1.set_ylabel(
    "Nombre total de femmes incluses",
    fontsize=10,
    fontweight="bold",
    color="#000",
)
ax1.tick_params(axis="y", labelcolor="#000")
ax1.set_yscale("log")  # Échelle logarithmique
ax1.set_ylim(10000, 2500000)

# Axe secondaire (Droite) : Nombre de cas de cancer
ax2 = ax1.twinx()
rects2 = ax2.bar(
    x + width / 2,
    cas_cancers,
    width,
    label="Cas de cancer invasif (Total N=18 368)",
    color="#97252E",
    edgecolor="black",
    alpha=0.85,
)
ax2.set_ylabel(
    "Nombre de cas de cancer du sein",
    fontsize=10,
    fontweight="bold",
    color="#97252E",
)
ax2.tick_params(axis="y", labelcolor="#97252E")
ax2.set_yscale("log")
ax2.set_ylim(100, 35000)

# Mise en forme globale
ax1.set_xticks(x)
ax1.set_xticklabels(ethnies, fontsize=10)
plt.title(
    "Répartition des effectifs bruts et cas de cancer par ethnie (BCSC v2)",
    fontsize=12,
    fontweight="bold",
    pad=15,
)

# Annotations des chiffres bruts au-dessus des barres
for bar in rects1:
    h = bar.get_height()
    ax1.text(
        bar.get_x() + bar.get_width() / 2,
        h * 1.15,
        f"{int(h):,}".replace(",", " "),
        ha="center",
        va="bottom",
        fontsize=8.5,
        color="#000",
        fontweight="bold",
    )

for bar in rects2:
    h = bar.get_height()
    ax2.text(
        bar.get_x() + bar.get_width() / 2,
        h * 1.15,
        f"{int(h):,}".replace(",", " "),
        ha="center",
        va="bottom",
        fontsize=8.5,
        color="#97252E",
        fontweight="bold",
    )

# Légende combinée
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(
    lines1 + lines2,
    labels1 + labels2,
    loc="upper right",
    frameon=True,
    fontsize=9.5,
)

ax1.grid(True, linestyle=":", alpha=0.4)
plt.tight_layout()
plt.savefig("BCSC_v2_Effectifs_Ethnies_Seul.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Style clean et académique
plt.style.use(
    "seaborn-v0_8-whitegrid"
    if "seaborn-v0_8-whitegrid" in plt.style.available
    else "default"
)
fig, ax = plt.subplots(figsize=(12, 5))

# Données brutes
categories = ["Témoins\n(Controls)", "Cas RE+\n(Estrogen Receptor +)", "Cas RE-\n(Estrogen Receptor -)", "Cas statut RE\ninconnu / autre"]
effectifs = [95567, 69501, 11500, 133384 - (69501 + 11500)] # 52 383 cas statut non spécifié
couleurs = ["#000", "#97252E", "#97252E", "#97252E"]

bars = ax.bar(categories, effectifs, color=couleurs, width=0.55, edgecolor="black", alpha=0.85)

# Axes et titres
ax.set_ylabel("Nombre de participantes", fontsize=10, fontweight="bold")
ax.set_title("Répartition de la cohorte d'apprentissage BCAC (N = 228 951)", fontsize=11, fontweight="bold", pad=15)
ax.set_ylim(0, 110000)

# Annotation des effectifs bruts et pourcentages
total_cohort = 228951
for bar in bars:
    height = bar.get_height()
    pct = (height / total_cohort) * 100
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        height + 2000,
        f"{int(height):,}\n({pct:.1f}%)".replace(",", " "),
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold"
    )

ax.grid(True, linestyle=":", alpha=0.5, axis="y")
plt.tight_layout()
plt.savefig("BCAC_Cohort_Distribution.png", dpi=300)
plt.show()
```
