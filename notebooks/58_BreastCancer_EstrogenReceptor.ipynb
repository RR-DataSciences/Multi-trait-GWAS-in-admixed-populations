## Estrogen Receptor Positive

```python
cdr_dataset = os.environ["WORKSPACE_CDR"]

# Requête SQL cherchant le concept 4167696 et tous ses sous-concepts (descendants)
query_er_pos = f"""
SELECT DISTINCT
    co.person_id,
    MIN(co.condition_start_date) AS er_pos_date,
    c.concept_id,
    c.concept_name
FROM `{cdr_dataset}.condition_occurrence` co
JOIN `{cdr_dataset}.concept_ancestor` ca 
    ON co.condition_concept_id = ca.descendant_concept_id
JOIN `{cdr_dataset}.concept` c 
    ON co.condition_concept_id = c.concept_id
WHERE ca.ancestor_concept_id = 4167696
GROUP BY co.person_id, c.concept_id, c.concept_name
"""

cmd = ["bq", "query", "--use_legacy_sql=false", "--format=csv", "--max_rows=500000", query_er_pos]
result = subprocess.run(cmd, capture_output=True, text=True, check=True)

df_er_pos = pd.read_csv(io.StringIO(result.stdout))
df_er_pos
```

```python
df_er_pos
```

```python
df.merge(df_er_pos, on='person_id', how='inner', indicator=True)
```
